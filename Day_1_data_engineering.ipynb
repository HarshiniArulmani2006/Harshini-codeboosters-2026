```python
import pandas as pd #step 1 pandas is the python library
print("Libraries imported Successfully!!")
```

```python
#load the dataset
df=pd.read_csv('/content/student_performance.csv')
```

```python

#View the dataset
print(df)
```

```python
#head shows the first 5 rows will be shown
df.head()
```

```python
#tail shows the last 5 rows will be shown
df.tail()
```

```python
#shape gives the Structure
print("Shape of the dataset:",df.shape)
print("Number of Students:",df.shape[0]) # 0 mentions row
print("Number of Columns:",df.shape[1]) # 1 mentions columns
```

```python
# View all the columns
print("Columns in the Datasets:")
print(df.columns.to_list()) #tolist gives format just not messy
```

```python
#dtypes used to check data types of each column
print("Data Types:")
print(df.dtypes)
```

```python
#describe shows the statistical value(mean,Variance,Standard deviation,min,max etc of the dataset )
df.describe()
```

```python
#isnull and sum used to check missing values
print("Missing Values in each columns:")
print(df.isnull().sum()) #isnull sees the null and sum returns the count of null value
print("\nTotal missing values:",df.isnull().sum().sum())
```

```python
#View Specific column
print("Student Names:")
print(df['name'])
```

```python
#View Unique departments or columns
print("Departments in the dataset:")
print(df['department'].unique())
print("\nHow many Students per department?")
print(df['department'].value_counts())
```

```python
#filters the students by department
cs_students=df[df['department']=='Computer Science']
print(f"Number of Computer Science Students:{len(cs_students)}")
print("\nComputer Science Students:")
print(cs_students[['name','math_score','programming_score','attendance_percentage']])
```

```python
#filter students with high attendance students(above 90%)
cs_students=df[(df['department']=='Computer Science')&(df['attendance_percentage']>90)]
print(f"Number of Computer Science Students:{len(cs_students)}")
print("\nComputer Science Students:")
print(cs_students[['name','math_score','programming_score','attendance_percentage']])
```

```python
#find the top performing student overall students with overall departments
df['total_score']=df['math_score']+df['science_score']+df['english_score']+df['programming_score']
top_students=df[df['total_score']/4>90]
print(top_students[['name','total_score']])
print("\n Top Students:",len(top_students))
```

```python
#Calculating Average score per department
department_average=df.groupby('department')['total_score'].mean().round(2)
print("Average Score per Department:")
print(department_average)
```

```python
#Average in descending order
department_average=df.groupby('department')['total_score'].mean().round(2).sort_values(ascending=False)
print("Average Score per Department:")
print(department_average)
```
