```python
import pandas as pd
import numpy as num
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import chromadb
print("All libraries imported successfully!")
print(f"ChromaDB version:{chromadb.__version__}")
```

```python
#Key Word Search
documents=[
    "ETL is used to clean and transform data",
    "A vehicle is a mode of transportation",
    "Cars and Trucks are popular automobiles",
    "SQL is used to query databases",
    "Machine Learning trains models on data",
]
query_keyword="vehicle"
print(f"KEYWORD SEARCH for:'{query_keyword}'")

for i,doc in enumerate(documents):
  if query_keyword.lower() in doc.lower():
    print(f"--> FOUND [doc_{i}]:{doc}")
  else:
    print(f"--> MISSED [doc_{i}]:{doc}")
  print()
  print()
  print("PROBLEM:doc_2 talks about 'Cars and trucks'-which are vehicles!")
  print("But keyword search MISSED it because it searched for the exact word 'vehicle'.")
```

```python
#for semantic search first have to do embeddings
#embedding
print("Loading Embedding Model...")
model=SentenceTransformer('all-MiniLM-L6-v2')
print("Embedding Model Loaded Successfully!!!")
print(f"Model Produces Vectors of Size:{model.get_sentence_embedding_dimension()} dimensions")
```

```python
#generate your first embedding
sentence="ETL is used to clean and transform data"
embedding=model.encode(sentence) #convert string to integer
print(f"Input Sentence:'{sentence}'")
print()
print(f"Embedding Type:{type(embedding)}")
print(f"Embedding shape:{embedding.shape}")
print(f"First 10 numbers:{embedding[:10].round(4)}")
print(f"Minimum Value:{embedding.min():.4f}")
print(f"Maximum Value:{embedding.max():.4f}")
```

```python
#semantic search
documents=[
    "ETL is used to clean and transform data",  # doc_0
    "A vehicle is a mode of transportation",    # doc_1
    "Cars and trucks are popular automobiles",  # doc_2
    "SQL is used to query databases",           # doc_3
    "Machine learning trains models on data",   # doc_4
]
query="vehicle"
model=SentenceTransformer('all-MiniLM-L6-v2')
doc_embeddings=model.encode(documents)
query_embedding=model.encode([query])
similarities=cosine_similarity(query_embedding,doc_embeddings)[0]
print(f"SEMANTIC SEARCH for:'{query}'")
threshold=0.20    #similarities like accuracy in output as like temperature
for i,score in enumerate(similarities):
    if score>threshold:
        print(f"FOUND [doc_{i}] (score={score:.3f}):{documents[i]}")
    else:
        print(f"MISSED [doc_{i}] (score={score:.3f}):{documents[i]}")
```

```python
sentences=[
         "ETL is used to clean and transform data",
         "Data Transformation is a key pipeline step",
         "The sky is blue and clouds are white",
]
embeddings=model.encode(sentences)
print(f'Number of Sentences:{len(sentences)}')
print(f'Shape of Embeddings array:{embeddings.shape}')
print()
print(f"Each Row is one sentence's embeddings:")
for i,sent in enumerate(sentences):
  print(f"--> Sentence{i}:shape={embeddings[i].shape},first 5 values={embeddings[i][:5]}")
```

```python
# my own sentence
your_sentences=[
  "Oryza sativa is the botanical name of Rice",
  "Tamil Nadu is the popular place for heritage",
  "Lysosomes are the Sucidal bags of the cell",
]
your_embeddings=model.encode(your_sentences)
sim_your_first=cosine_similarity([your_embeddings[0]],[your_embeddings[1]])
sim_your_second=cosine_similarity([your_embeddings[0]],[your_embeddings[2]])
print(f"EXPERIMENT RESULTS")
print(f"Sentence A:'{your_sentences[0]}'")
print(f"Sentence B:'{your_sentences[1]}'")
print(f"Sentence C:'{your_sentences[2]}'")
print(f"Similarity(A vs B):{sim_your_first[0][0]:.4f}")
print(f"Similarity(A vs C):{sim_your_second[0][0]:.4f}")
print()
```

```python
#going to write query
chroma_client=chromadb.Client()
collection=chroma_client.get_or_create_collection("first_client")
print("ChromaDB client created (in memory mode)")
print(f"Collection name:first_client")
print(f"Documents in Collection:{collection.count()}")
```

```python
sample_docs=[
    "ETL stands for Extract Transform Load-the core data engineering progress",
    "SQL SELECT statements retrieves data from databases tables",
    "Machine Learning models learn patterns from training data",
    "Python Pandas library is used for data manupulation and cleaning",
    "Neural Networks are inspired by how the human brain works",
]
sample_ids=["doc001","doc002","doc003","doc004","doc005"]
sample_metadatas=[{"subject":"Data Engineering","topic":"ETL"},
 {"subject":"Data Engineering","topic":"SQL"},
 {"subject":"Machine Learning","topic":"ML Basics"},
 {"subject":"Python","topic":"Pandas"},
 {"subject":"Machine Learning","topic":"Neural Networks"},
]
collection.add(
    documents=sample_docs,
    ids=sample_ids,
    metadatas=sample_metadatas,
    )
print(f"Documents added to collections!")
print(f"Total documents now in collections:{collection.count()}")
```

```python
query="How do I clean and prepare data?"
results=collection.query(query_texts=query,n_results=3)
print("RESULT KEYS AVAILABLE: ")
print(list(results.keys()))
```

```python
print(f"Query':'{query}'")
matched_docs=results["documents"][0]
matched_ids=results["ids"][0]
matched_distances=results["distances"][0]
matched_metadatas=results["metadatas"][0]
for rank,(doc,doc_id,dist,meta)in enumerate(zip(matched_docs,matched_ids,matched_distances,matched_metadatas)):
  print(f"Rank:{rank}|ID:{doc_id}|Distance:{dist:.4f}")
  print(f"-->Subjects:{meta['subject']}|Topics:{meta['topic']}")
  print(f"-->Document:{doc}")
  print()

print("NOTICE:The results are about ETL and Pandas-exactly what 'clean and prepare data'")
print("Semantic search found them even though the words are different")

```

```python
filtered_query="How do computers learn from examples?"
filtered_results=collection.query(
    query_texts=filtered_query,
    n_results=3,
    where={"subject":"Machine Learning"},
)
print(f"Filtered Query:'{filtered_query}'")
print("Filter:Only Machine Learning Documents")
for rank,(doc,dist,meta)in enumerate(zip(
    filtered_results["documents"][0],
    filtered_results["distances"][0],
    filtered_results["metadatas"][0]),start=1):
  print(f"Rank{rank}|Distance:{dist:.4f}|Subject:{meta['subject']}")
  print(f"-->Document{doc}")
  print()

print("Notice:Only ML documents appear,even though ETL and Pandas might be related")
```
